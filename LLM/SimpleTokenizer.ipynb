```python
import re

class SimpleTokenizer:
    def __init__(self, vocabulary):
        self.word_id = vocabulary
        self.id_word = {i:w for w,i in vocabulary.items()}
    def encode(self, text):
        preprocessed = [w.strip() for w in re.split(r'[.,?_!"()\']|--|\s', text) if w.strip()]
        return [self.word_id[w] for w in preprocessed]
    def decode(self, ids):
        text = " ".join(self.id_word[i] for i in ids)        
        return re.sub(r'\s+([.,?!"()\'])', r'\1', text) 


class SimpleTokenizerV2:
    def __init__(self, vocabulary):
        self.word_id = vocabulary
        self.id_word = {i:w for w,i in vocabulary.items()}
    def encode(self, text):
        preprocessed = [w.strip() for w in re.split(r'[.,?_!"()\']|--|\s', text) if w.strip()]
        return [self.word_id[w] if w in self.word_id else self.word_id['<|unk|>'] for w in preprocessed]
    def decode(self, ids):
        text = " ".join(self.id_word[i] for i in ids)        
        return re.sub(r'\s+([.,?!"()\'])', r'\1', text)         
```

```python
filepath = "the-verdict.txt"
with open(filepath, "r", encoding="utf-8") as f:
    raw_text = f.read()
```

```python
raw_text_preprocessed =  [w.strip() for w in re.split(r'[.,?_!"()\']|--|\s', raw_text) if w.strip()]
raw_text_preprocessed = sorted(set(raw_text_preprocessed))
raw_text_preprocessed.extend(['<|unk|>','<|endoftext|>'])
```

```python
vocabulary = {w: i  for i,w in enumerate(raw_text_preprocessed)}
```

```python
tokenizer = SimpleTokenizer(vocabulary)
```

```python
print('test 1')
```

```python
text1 = """"It's last time he painted, you know,
        "Mrs. Gisburn said with pardonable pride."""
```

```python
ids = tokenizer.encode(text1)
```

```python
print(ids)
```

```python
print(tokenizer.decode(ids))
```
