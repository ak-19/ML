```python
raw_text_preprocessed =  [w.strip() for w in re.split(r'([,.:;?_!"()\']|--|\s)', raw_text) if w.strip()]
```

```python
print(len(raw_text_preprocessed))
```

```python
print(raw_text_preprocessed[:30])
```

```python
all_words = sorted(set(raw_text_preprocessed))
```

```python
len(all_words)
```

```python
token_id = {w: i  for i,w in enumerate(all_words)}
```

```python
id_token = {i: w  for w, i in token_id.items()}
```
